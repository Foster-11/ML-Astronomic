```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('sdss_sample.csv')
```

```python
features = ['u', 'g', 'r', 'i', 'z']
X = df[features]
y = df['redshift']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
scaler = MinMaxScaler()

scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
kmn = KMeans(n_clusters = 3)

kmn.fit(X_train, )
clusters = kmn.predict(X_test)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfica 1 — Clusters de KMeans
ax1.scatter(X_test[:, 0], X_test[:, 1], c=clusters, cmap='viridis', alpha=0.5)
ax1.set_title('Clusters KMeans')
ax1.set_xlabel('u')
ax1.set_ylabel('g')

# Gráfica 2 — Clases reales
y_test_encoded = LabelEncoder().fit_transform(y_test)
ax2.scatter(X_test[:, 0], X_test[:, 1], c=y_test_encoded, cmap='viridis', alpha=0.5)
ax2.set_title('Clases Reales')
ax2.set_xlabel('u')
ax2.set_ylabel('g')

plt.tight_layout()
plt.savefig('outputs/kmeans_clusters.png', dpi=150)
plt.show()
```
